```python
import pandas
```

```python
hogwarts = pandas.read_csv('hogwarts_dataset.csv', sep=";")
```

```python
from sklearn.tree import DecisionTreeClassifier # Import Decision Tree Classifier
from sklearn.model_selection import train_test_split # Import train_test_split function
from sklearn import metrics #Import scikit-learn metrics module for accuracy calculation
```

```python
hogwarts['gender'] = hogwarts['gender'].replace({'Male': 0, 'Female': 1})
```

```python
hogwarts
```

```python
hogwarts = pandas.get_dummies(hogwarts,columns=["hair_colour"], drop_first=False)
```

```python
hogwarts
```

```python
hogwarts
```

```python
#split dataset in features and target variable

X = hogwarts[['lenght_wand','max_broom_speed','gender','hair_colour_black','hair_colour_brown','hair_colour_white']] # Features

#X = hogwarts[hogwarts.columns.difference(['name','house'])]
y = hogwarts.house
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=2)
```

```python
X_train
```

```python
y_train
```

```python
# Create Decision Tree classifer object
clf = DecisionTreeClassifier()
```

```python
clf = clf.fit (X_train,y_train)
```

```python
y_pred = clf.predict(X_test)

```

```python
X_test
```

```python
y_test
```

```python
y_pred
```

```python
# Model Accuracy, how often is the classifier correct?
print("Accuracy:",metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn.tree import export_graphviz
from six import StringIO
from IPython.display import Image
```

```python
import pydotplus
```

```python
hogwarts_columns = X_train.columns.to_list()

dot_data = StringIO()
export_graphviz(clf, out_file=dot_data,
                filled=True, rounded=False,
                special_characters=True,feature_names = hogwarts_columns ,class_names=['Gryffindor', 'Slytherin'])
graph = pydotplus.graph_from_dot_data(dot_data.getvalue())
graph.write_png('graf.png')
Image(graph.create_png())
```
